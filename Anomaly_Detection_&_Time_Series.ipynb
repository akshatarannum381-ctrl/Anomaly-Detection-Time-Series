1. What is Anomaly Detection? Explain its types (point, contextual, and
collective anomalies) with examples.
- Anomaly Detection is a technique in data analysis and machine learning used to identify data points, events, or patterns that do not conform to expected behavior. These unusual observations are often called outliers, novelties, or exceptions. Anomaly detection is widely used in fraud detection, network security, fault detection, and monitoring industrial machines.

1. Point Anomalies (or Global Outliers)

Definition: A single data point is considered anomalous if it deviates significantly from the rest of the data.

Example:

In a dataset of daily temperatures: [30, 32, 31, 33, 29, 45, 32], the value 45°C is unusually high and is a point anomaly.

Credit card transactions: A sudden $10,000 purchase on a card that normally sees $10–$50 transactions.

Use Case: Fraud detection, network intrusion detection, sensor monitoring.

2. Contextual Anomalies (or Conditional Anomalies)

Definition: A data point is anomalous only within a specific context, usually considering time, location, or other contextual features.

Example:

Temperature anomaly: A temperature of 5°C might be normal in winter but anomalous in summer.

Stock prices: A sudden drop in stock price is unusual if it happens during a stable market, but may be normal during a crash.

Use Case: Seasonal monitoring, time-series analysis, weather forecasting.

3. Collective Anomalies

Definition: A group of data points is anomalous together, even if individual points are normal. The anomaly is in the pattern of the group rather than a single point.

Example:

Network traffic: A sudden burst of connections to a server (even if each connection looks normal individually) may indicate a DDoS attack.

Heartbeat monitoring: A series of heart rate measurements forming an unusual pattern may indicate arrhythmia, even if each reading is within normal range.

Use Case: Detecting unusual sequences or patterns, fraud rings, system failures.

3.  What are the key components of a Time Series? Explain each with one
example.
- A time series is a sequence of data points collected over time at regular intervals. Understanding its key components is crucial for analysis, forecasting, and modeling. The main components of a time series are:

1. Trend (T)

Definition: The long-term movement or direction in the data over time. It shows whether the series is generally increasing, decreasing, or stable.

Example:

The annual sales of an e-commerce company steadily increasing from $1M to $5M over 5 years.

Use in Analysis: Helps in identifying growth patterns and making long-term forecasts.

2. Seasonal Component (S)

Definition: Regular, repeating patterns in the data over fixed periods (daily, weekly, monthly, yearly). Seasonality is predictable.

Example:

Ice cream sales increase every summer and decrease in winter.

Website traffic spikes every Monday due to weekly newsletter campaigns.

Use in Analysis: Important for adjusting predictions for regular fluctuations.

3. Cyclical Component (C)

Definition: Fluctuations around the trend that occur over irregular intervals, usually influenced by economic or business cycles. Cycles are longer-term than seasonal patterns and not fixed.

Example:

The stock market goes through bull and bear cycles lasting several years.

GDP growth rates rise and fall due to economic cycles.

Use in Analysis: Helps in understanding medium-to-long-term patterns influenced by external factors.

4. Irregular/Random Component (I)

Definition: Unpredictable, random variations that cannot be attributed to trend, seasonality, or cycles. Also called noise.

Example:

A sudden drop in sales due to an unexpected flood or natural disaster.

Stock price fluctuations caused by unexpected news.

Use in Analysis: Usually removed or smoothed out to better analyze underlying patterns.

Question 6: Load a time series dataset (e.g., AirPassengers), plot the original series,
and decompose it into trend, seasonality, and residual components

```python

import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose


from statsmodels.datasets import get_rdataset


data = get_rdataset("AirPassengers").data


data['time'] = pd.to_datetime(data['time'], format='%Y')

data.index = pd.date_range(start='1949-01', periods=len(data), freq='M')


data.rename(columns={'value': 'Passengers'}, inplace=True)


plt.figure(figsize=(12,5))
plt.plot(data['Passengers'], color='blue')
plt.title('AirPassengers Time Series')
plt.xlabel('Year')
plt.ylabel('Number of Passengers')
plt.show()


decomposition = seasonal_decompose(data['Passengers'], model='multiplicative', period=12)


plt.figure(figsize=(12,9))

plt.subplot(411)
plt.plot(data['Passengers'], label='Original', color='blue')
plt.legend(loc='best')

plt.subplot(412)
plt.plot(decomposition.trend, label='Trend', color='red')
plt.legend(loc='best')

plt.subplot(413)
plt.plot(decomposition.seasonal, label='Seasonality', color='green')
plt.legend(loc='best')

plt.subplot(414)
plt.plot(decomposition.resid, label='Residuals', color='orange')
plt.legend(loc='best')

plt.tight_layout()
plt.show()
```

7. Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to
detect anomalies. Visualize the anomalies on a 2D scatter plot.


```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest


df = pd.read_csv('nyc_taxi.csv')


data = df[['fare_amount', 'trip_distance']]


data = data[(data['fare_amount'] > 0) & (data['trip_distance'] > 0)]


iso_forest = IsolationForest(contamination=0.01, random_state=42)

iso_forest.fit(data)


data['anomaly'] = iso_forest.predict(data)


normal = data[data['anomaly'] == 1]
anomalies = data[data['anomaly'] == -1]


plt.figure(figsize=(10,6))
plt.scatter(normal['trip_distance'], normal['fare_amount'], c='blue', label='Normal', alpha=0.6)
plt.scatter(anomalies['trip_distance'], anomalies['fare_amount'], c='red', label='Anomaly', alpha=0.8)
plt.xlabel('Trip Distance (miles)')
plt.ylabel('Fare Amount ($)')
plt.title('Isolation Forest Anomaly Detection - NYC Taxi Fare')
plt.legend()
plt.show()
```

8. Train a SARIMA model on the monthly airline passengers dataset.
Forecast the next 12 months and visualize the results.

```python

import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.datasets import get_rdataset


data = get_rdataset("AirPassengers").data
data['time'] = pd.date_range(start='1949-01', periods=len(data), freq='M')
data.set_index('time', inplace=True)
data.rename(columns={'value': 'Passengers'}, inplace=True)


plt.figure(figsize=(12,5))
plt.plot(data['Passengers'], label='Original')
plt.title('Monthly AirPassengers Time Series')
plt.xlabel('Year')
plt.ylabel('Number of Passengers')
plt.legend()
plt.show()


sarima_model = SARIMAX(data['Passengers'],
                       order=(1,1,1),
                       seasonal_order=(1,1,1,12),
                       enforce_stationarity=False,
                       enforce_invertibility=False)

sarima_result = sarima_model.fit(disp=False)
print(sarima_result.summary())


forecast = sarima_result.get_forecast(steps=12)
forecast_index = pd.date_range(start=data.index[-1] + pd.offsets.MonthBegin(1), periods=12, freq='M')
forecast_values = forecast.predicted_mean
conf_int = forecast.conf_int()


plt.figure(figsize=(12,6))
plt.plot(data['Passengers'], label='Original')
plt.plot(forecast_index, forecast_values, color='red', marker='o', label='Forecast (Next 12 Months)')
plt.fill_between(forecast_index, conf_int['lower Passengers'], conf_int['upper Passengers'], color='pink', alpha=0.3)
plt.title('SARIMA Forecast of AirPassengers')
plt.xlabel('Year')
plt.ylabel('Number of Passengers')
plt.legend()
plt.show()
```

9. Apply Local Outlier Factor (LOF) on any numerical dataset to detect
anomalies and visualize them using matplotlib.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor


np.random.seed(42)

X_normal = np.random.randn(200, 2)

X_outliers = np.random.uniform(low=-6, high=6, size=(10,2))

X = np.vstack([X_normal, X_outliers])
df = pd.DataFrame(X, columns=['Feature1', 'Feature2'])


lof = LocalOutlierFactor(n_neighbors=20, contamination=0.05)
df['anomaly'] = lof.fit_predict(df[['Feature1', 'Feature2']])


normal = df[df['anomaly'] == 1]
anomalies = df[df['anomaly'] == -1]


plt.figure(figsize=(10,6))
plt.scatter(normal['Feature1'], normal['Feature2'], c='blue', label='Normal', alpha=0.6)
plt.scatter(anomalies['Feature1'], anomalies['Feature2'], c='red', label='Anomaly', alpha=0.8)
plt.xlabel('Feature1')
plt.ylabel('Feature2')
plt.title('Local Outlier Factor (LOF) - Anomaly Detection')
plt.legend()
plt.show()
```
